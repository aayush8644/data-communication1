```python
import numpy as np
import matplotlib.pyplot as plt

# Time vector
t = np.arange(0, 1, 0.0001)

# User inputs
Am = float(input("Message amplitude (Am): "))
Ac = float(input("Carrier amplitude (Ac): "))
Fm = float(input("Message frequency (Fm): "))
Fc = float(input("Carrier frequency (Fc): "))

# Message signal
Mt = Am * np.cos(2 * np.pi * Fm * t)

# Carrier signal
Ct = Ac * np.cos(2 * np.pi * Fc * t)

# Modulation Index
Ma = Am / Ac

# Amplitude Sensitivity
Ka = Ma / Am

# AM Signal
S = Ac * (1 + Ka * Mt) * np.cos(2 * np.pi * Fc * t)

# Plotting
plt.figure(figsize=(10, 8))

# Message Signal
plt.subplot(3, 1, 1)
plt.plot(t, Mt, linewidth=2)
plt.title("Message Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# Carrier Signal
plt.subplot(3, 1, 2)
plt.plot(t, Ct, 'r', linewidth=2)
plt.title("Carrier Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# AM Signal
plt.subplot(3, 1, 3)
plt.plot(t, S, 'g', linewidth=2)
plt.title("Amplitude Modulated (AM) Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()

# Display Modulation Index
print("Modulation Index (Ma):", Ma)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Time vector
t = np.arange(0, 1, 0.001)

# User inputs
Am = float(input("Message amplitude (Am): "))
Ac = float(input("Carrier amplitude (Ac): "))
Fm = float(input("Message frequency (Fm): "))
Fc = float(input("Carrier frequency (Fc): "))

# Message signal
Mt = Am * np.cos(2 * np.pi * Fm * t)

# Carrier signal
Ct = Ac * np.cos(2 * np.pi * Fc * t)

# DSB-SC Modulated Signal
DSB_SC = Mt * Ct

# Plotting
plt.figure(figsize=(10, 8))

# Message Signal
plt.subplot(3, 1, 1)
plt.plot(t, Mt, linewidth=2)
plt.title("Message Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# Carrier Signal
plt.subplot(3, 1, 2)
plt.plot(t, Ct, 'r', linewidth=2)
plt.title("Carrier Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

# DSB-SC Signal
plt.subplot(3, 1, 3)
plt.plot(t, DSB_SC, 'g', linewidth=2)
plt.title("DSB-SC Modulated Signal")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Input bits
x = [1, 1, 0, 0, 1, 0, 1, 0]

f = 1          # Base frequency
Tb = 1         # Bit duration
samples = 100  # Samples per bit

ask = []
fsk = []
psk = []

for bit in x:
    t = np.linspace(0, Tb, samples)

    # ASK
    ask_signal = bit * np.sin(2 * np.pi * f * t)
    ask.extend(ask_signal)

    # FSK
    freq = 2 * f if bit == 1 else f
    fsk_signal = np.sin(2 * np.pi * freq * t)
    fsk.extend(fsk_signal)

    # PSK
    phase = 0 if bit == 1 else np.pi
    psk_signal = np.sin(2 * np.pi * f * t + phase)
    psk.extend(psk_signal)

# Time axis
t_total = np.linspace(0, len(x) * Tb, len(ask))

# Plotting
plt.figure(figsize=(10, 8))

plt.subplot(3, 1, 1)
plt.plot(t_total, ask)
plt.title("ASK Modulation")
plt.grid(True)

plt.subplot(3, 1, 2)
plt.plot(t_total, fsk)
plt.title("FSK Modulation")
plt.grid(True)

plt.subplot(3, 1, 3)
plt.plot(t_total, psk)
plt.title("PSK Modulation")
plt.grid(True)

plt.tight_layout()
plt.show()
```
